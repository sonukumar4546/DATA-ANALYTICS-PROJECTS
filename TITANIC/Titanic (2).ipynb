# 📄 Titanic Dataset — Description

The Titanic dataset contains information about passengers aboard the RMS Titanic and is commonly used for data analysis and machine-learning classification tasks.

---

## 📌 Column-wise Description

| Column Name    | Type      | Description |
|----------------|-----------|-------------|
| **PassengerId** | Integer | Unique identifier assigned to each passenger. |
| **Survived** | Integer (0/1) | Survival indicator — **1 = Survived**, **0 = Did not survive**. |
| **Pclass** | Integer (1/2/3) | Passenger ticket class — **1 = Upper class**, **2 = Middle class**, **3 = Lower class**. |
| **Name** | String | Full passenger name, including title (Mr, Mrs, Miss, etc.). |
| **Sex** | Integer (0/1) | Encoded gender — **0 = Male**, **1 = Female**. |
| **Age** | Float | Passenger’s age in years. Contains some missing values. |
| **SibSp** | Integer | Number of siblings/spouses aboard the ship. |
| **Parch** | Integer | Number of parents/children aboard the ship. |
| **Ticket** | String | Ticket number of the passenger. |
| **Fare** | Float | Amount paid for the ticket. |
| **Embarked** | Integer (0/1/2) | Port of embarkation — **0 = Southampton**, **1 = Cherbourg**, **2 = Queenstown**. |
| **has_Cabin** | Integer (0/1) | Indicates whether cabin information is available — **1 = Yes**, **0 = No**. |

---

### FILL EMPTY AGE WITH THE MEDAIN AGE

```python
df['Age']=df['Age'].fillna(df['Age'].median())
```
