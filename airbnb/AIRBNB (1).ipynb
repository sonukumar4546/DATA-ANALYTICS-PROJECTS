# 🏠 Airbnb Dataset Description

## 📋 Overview
This dataset contains **14,456 entries** of Airbnb listings with detailed information about properties, hosts, reviews, and pricing.  
It is structured as a tabular dataset with **13 columns**, mixing numerical, categorical, and textual data.

---

## 📂 Columns and Their Meaning

| Column        | Type    | Description                                                                 |
|---------------|---------|-----------------------------------------------------------------------------|
| `id`          | int64   | Unique identifier for each listing                                          |
| `name`        | object  | Title or name of the listing                                                |
| `rating`      | object  | Average rating score (e.g., 4.71, 5.00, or “New”)                          |
| `reviews`     | object  | Number of reviews received (may contain missing values)                     |
| `host_name`   | object  | Name of the host, often with property type (e.g., “Entire villa hosted by…”)|
| `host_id`     | float64 | Unique identifier for the host                                              |
| `address`     | object  | Location details (city, region, country)                                    |
| `features`    | object  | Property features (guests, bedrooms, beds, bathrooms)                       |
| `amenities`   | object  | List of amenities (e.g., kitchen, mountain view, Wi-Fi, lake access)        |
| `safety_rules`| object  | Safety guidelines (e.g., COVID-19 practices, pets allowed, smoking rules)   |
| `hourse_rules`| object  | House rules (check-in/out times, restrictions)                              |
| `img_links`   | object  | URL links to property images                                                |
| `price`       | object  | Price per night (currency formatted, e.g., ₹8,078 per night)                |

```python
df['bathroom']=df['features'].str.split(',').str[-1]
```

```python
df['bathroom'] = df['bathroom'].str.replace('bathrooms', 'bathroom', regex=False)
df['bathroom'] = df['bathroom'].str.replace('Private ', '', regex=False)
df['bathroom'] = df['bathroom'].str.replace('Shared ', '', regex=False)
```

```python
df['bathroom'] = df['bathroom'].str.replace('Toilet', '0.5 bathroom', regex=False)
```

## no of bed extract

```python
df['no_of_bedroom'] = df['features'].str.extract(r'(\d+)\s*(?:bed|beds)', expand=False)
df['no_of_bedroom'] = pd.to_numeric(df['no_of_bedroom'], errors='coerce')
df['no_of_bedroom'] = df['no_of_bedroom'].fillna(0).astype(int)
```

```python
import matplotlib.pyplot as plt

# Get the top 10 countries by average price
top_10_countries = df.groupby('country')['Price'].mean().nlargest(10).index

# Filter the DataFrame to include only these top countries
df_top_countries = df[df['country'].isin(top_10_countries)]

# Create a box plot to visualize the price distribution for these countries
plt.figure(figsize=(15, 8))
sns.boxplot(x='country', y='Price', data=df_top_countries, palette='viridis')
plt.title('Price Distribution by Top 10 Countries (by Average Price)')
plt.xlabel('Country')
plt.ylabel('Price')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
