## 📘 Overview
The **Songs Dataset** contains metadata of **41,355 songs**, including information about song titles, artists, durations, and languages. It serves as a useful resource for music analysis, recommendation systems, and multilingual content insights.

---

## 📊 Dataset Summary

| Feature | Description | Data Type | Example | Unique Values |
|----------|--------------|------------|-----------|----------------|
| **name** | Title of the song | `object` | `Dil - E - Nadan Tujhe` | 31,534 |
| **singer** | Name(s) of the singer(s), sometimes multiple separated by `|` | `object` | `Chitra Singh | Jagjit Singh` | 6,196 |
| **singer_id** | Unique artist identifier(s) corresponding to the singers | `object` | `/artist/jagjeet-singh-1` | 6,580 |
| **duration** | Duration of the song in `mm:ss` format | `object` | `05:00` | 1,309 |
| **link** | URL path or identifier to access the song details | `object` | `/dil-e-nadan-tujhe-3` | 36,361 |
| **language** | Language of the song | `object` | `Urdu`, `Telugu`, `Hindi`, etc. | 16 |

# DATA CLEANING

### convert duration in sec

```python
lsta=[]
for i in df['duration']:
  if len(i.split(':'))==3:
    hour=int(i.split(':')[0])
    min=int(i.split(':')[1])
    sec=int (i.split(':')[2])
    duration=(hour*3600+min*60+sec)
    lsta.append(duration)
  elif len(i.split(':'))==2:
    min=int(i.split(':')[0])
    sec=int (i.split(':')[1])
    duration=(min*60+sec)
    lsta.append(duration)
```

```python
df['duration_in_sec']=lsta
```

```python
#del duration_in_sec
del lsta
```

### total duration time in sec  in a df



```python
counta=0
for i in df['duration']:
  if len(i.split(':'))==3:
    counta += int(i.split(':')[0])*3600
    counta += int(i.split(':')[1])*60
    counta += int(i.split(':')[2])
  elif len(i.split(':'))==2:
    counta += int(i.split(':')[0])*60
    counta += int(i.split(':')[1])
print(counta)
```
